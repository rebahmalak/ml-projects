```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([50,65,80,100,120,150]) #house sizes in square meters (feature/input)
y = np.array([110,130,155,210,230,310]) #house prices (target/label/output)
```

```python
plt.scatter(x,y)
plt.xlabel("house size (m^2)")
plt.ylabel("house price (*100$)")
plt.show()
```

```python
w=4
b=0
predictions = w*x+b
print(predictions)
```

```python
def visualization (x,y,predictions):
  plt.scatter(x,y)
  plt.plot(x,predictions)
  plt.xlabel("house size (m^2)")
  plt.ylabel("house price (*100$)")
  plt.title("house price prediction")
  plt.show()
visualization (x,y,predictions)
```

```python
house_size= 90
predicted_price = w * house_size + b
print("predicted price is :", predicted_price)
```

```python
def compute_cost(x,y,predictions):
    m = len(x)
    errors = predictions - y
    squared_errors = errors**2
    cost = (1/(2*m))* np.sum(squared_errors)
    return cost
print(compute_cost(x,y,predictions))
```

```python
w = 0
b = 0
```

```python
def compute_gradient(x,y,w,b):
  m = len(x)
  dj_dw = 0
  dj_db = 0
  for i in range(m):
    prediction = w * x[i] + b
    error = prediction - y[i]

    dj_dw += error * x[i]
    dj_db += error

  dj_dw = dj_dw / m
  dj_db = dj_db / m

  return dj_dw , dj_db
```

```python
alpha = 0.00001
```

```python
iterations = 1000
for i in range(iterations):
  dj_dw , dj_db =  compute_gradient(x,y,w,b)
  w = w- alpha * dj_dw
  b = b- alpha * dj_db
print("learned w:",w)
print("learned b:",b)

```

```python
predictions = w*x+b
print("pridections:", predictions)
```

```python
print(compute_cost(x,y,predictions))
```

```python
visualization (x,y,predictions)
```

```python
house_size= 90
predicted_price = w * house_size + b
print("predicted price is :", predicted_price)
```
